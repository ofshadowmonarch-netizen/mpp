```python
# descrete fourier transform :
import math

def simple_dft(signal):
    N = len(signal)
    F = []

    for k in range(N):
        total = 0

        for n in range(N):
            theta = (2 * math.pi * k * n) / N
            e_power = math.e ** (-1j * theta)
            total += signal[n] * e_power

        F.append(total)

    return F

input_data = [1, 2, 3, 4]
answer = simple_dft(input_data)

print(answer)
```

```python
# Finite Difference Method
import matplotlib.pyplot as plt
import numpy as np

n = 1000
x_start, x_end = 0, 2 * np.pi
h = (x_end - x_start) / n

p = 0
q = 0

X = np.linspace(x_start + h, x_end, n)
A = np.zeros((n, n))
R = np.zeros(n)

for i in range(n):
    A[i, i] = -2 + q * (h**2)
    R[i] = 5 * (h**2)

    if i > 0:
        A[i, i - 1] = 1 - (h * p / 2)
    if i < n - 1:
        A[i, i + 1] = 1 + (h * p / 2)

Y = np.linalg.solve(A, R)

plt.plot(X, Y, label="Solution", color="purple")
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Finite Difference Method Solution")
plt.grid(True)
plt.show()
```
